```python
s=7
guess=int(input("請輸: "))
while guess != s:
  if guess>s:
    print("太大了")
  else:
    print("太小了")
  guess=int(input("再猜"))
print("猜中了")
```

```python
import random
day=1
max_day=3
win=False
while day<=max_day and not win:
  lucky=random.randint(1,50)
  guess=int(input(f"今天第{day}天,請輸入1-50數字: "))
  print(f"今天的幸運數字是{lucky}")
  if guess == lucky:
    print("你中獎了")
    win=True
  else:
    print("你沒中獎，明天再來")
    day+=1
if not win:
  print("很遺憾3天都沒中獎，下次再試試吧！")
```
